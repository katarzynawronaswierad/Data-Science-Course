```python
import os

project_name = "sales_analysis"
```

```python
# Zdefiniowanie struktury projektu
project_structure = {
	'data': ['raw', 'processed'],
	'notebooks': [],
	'src': [],
	'models': [],
	'reports': []
}

def create_project_structure(project_name, structure):
	# Tworzenie głównego folderu
	os.makedirs(project_name, exist_ok=True)
	print(f"✓ Utworzono projekt: {project_name}/")
```

```python
# Tworzenie podfolderów
for folder, subfolders in project_structure.items():
	folder_path = os.path.join(project_name, folder)
	os.makedirs(folder_path, exist_ok=True)
	print(f" ├── {folder}/")
		
	for subfolder in subfolders:
		subfolder_path = os.path.join(folder_path, subfolder)
		os.makedirs(subfolder_path, exist_ok=True)
		print(f" │ ├── {subfolder}/")
```

```python
# Tworzenie plików konfiguracyjnych
files_to_create = {
'README.md': f'# {project_name}\n\nOpis projektu Data Science.',
'.gitignore':
'data/raw/\nmodels/trained/\n*.pyc\n__pycache__/\n.ipynb_checkpoints/',
'requirements.txt': 'pandas>=1.5.0\nnumpy>=1.23.0\nmatplotlib>=3.6.0\nscikit-learn>=1.2.0',
'config.py': '# Konfiguracja projektu\nDATA_PATH = "data/"\nMODEL_PATH = "models/"'
}
for filename, content in files_to_create.items():
    filepath = os.path.join(project_name, filename)
    with open(filepath, 'w') as f:
        f.write(content)
    print(f" ├── {filename}")
print(f"\n✓ Struktura projektu '{project_name}' utworzona!")
```
