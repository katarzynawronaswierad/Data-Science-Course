```python
import matplotlib.pyplot as plt

urzadzenia = ["Laptop", "Desktop", "Server"]
zuzycie_wh = [30, 150, 400]
kolory = ["#2E86AB", "#F6AE2D", "#D1495B"]   # niebieski, żółty, czerwony

fig, ax = plt.subplots(figsize=(9, 5.5))
slupki = ax.bar(urzadzenia, zuzycie_wh, color=kolory, edgecolor="black", linewidth=0.8)

# Tytuł i etykiety osi
ax.set_title("Zużycie energii przez urządzenia", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Typ urządzenia", fontsize=12)
ax.set_ylabel("Zużycie energii [Wh]", fontsize=12)

# Wartości nad słupkami
ax.bar_label(slupki, labels=[f"{v} Wh" for v in zuzycie_wh], padding=4, fontsize=11, fontweight="bold")

# Siatka i zakres osi Y
ax.set_ylim(0, max(zuzycie_wh) * 1.15)
ax.yaxis.grid(True, linestyle="--", alpha=0.6)
ax.set_axisbelow(True)                       # siatka pod słupkami

# Legenda (każde urządzenie ma swój kolor)
ax.legend(slupki, urzadzenia, title="Urządzenie", loc="upper left")

# Usunięcie górnej i prawej ramki
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
plt.tight_layout()
plt.savefig("energy_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
```
