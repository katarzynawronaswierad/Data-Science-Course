```python
import numpy as np
# Wyniki testów studentów
scores = np.array([45, 78, 92, 34, 88, 56, 91, 23, 67, 85])
# Warunek zwraca tablicę boolean
high_scores_mask = scores >= 80
print("Maska (scores >= 80):", high_scores_mask)
# Użycie maski do filtrowania
high_scores = scores[high_scores_mask]
print("Wysokie wyniki:", high_scores) # [92 88 91 85]
# Krótsza wersja
print("Niskie wyniki (<50):", scores[scores < 50]) # [45 34 23]
# Złożone warunki — AND (&), OR (|), NOT (~)
medium_scores = scores[(scores >= 60) & (scores < 80)]
print("Średnie wyniki [60-80):", medium_scores) # [78 67]
# Modyfikacja przez warunek
scores_copy = scores.copy()
scores_copy[scores_copy < 50] = 50
print("Po podniesieniu min do 50:", scores_copy)
# Zliczanie
num_passed = np.sum(scores >= 60) # True = 1, False = 0
print(f"\nLiczba zdających (>=60): {num_passed}")
print(f"Procent zdających: {num_passed / len(scores) * 100:.1f}%")
```

```python
import numpy as np
exam_scores = np.array([
[85, 78, 92, 88, 76],
[90, 85, 88, 92, 89],
[78, 72, 85, 80, 75],
[92, 88, 95, 90, 93],
[67, 70, 75, 72, 68]
])

student_means = np.mean(exam_scores, axis=1)
exam_means = np.mean(exam_scores, axis=0)

# Wykres "słupkowy" w trybie tekstowym (1 znak '█' = 2 punkty)
print("=== Średnie per student (█ = 2 punkty) ===")
for i, srednia in enumerate(student_means):
  bar = "█" * int(srednia // 2)
  print(f"Student {i+1}: {bar} {srednia:.1f}")

print("\n=== Średnie per egzamin (█ = 2 punkty) ===")
for i, srednia in enumerate(exam_means):
  bar = "█" * int(srednia // 2)
  print(f"Egzamin {i+1}: {bar} {srednia:.1f}")

# "Heatmapa" tekstowa — używamy symboli do oznaczenia zakresu
print("\n=== Heatmapa ocen (·<70, ·70–79, ■80–89, █>=90) ===")
def symbol(x):
  if x >= 90:
    return "█"
  elif x >= 80:
    return "■"
  elif x >= 70:
    return "·"
  else:
    return " "

print(" E1 E2 E3 E4 E5")

for i in range(exam_scores.shape[0]):
  linia = f"Student {i+1}: "
  for j in range(exam_scores.shape[1]):
    linia += f" {symbol(exam_scores[i, j])}{exam_scores[i, j]:2d}"
  print(linia)
```

```python
import numpy as np
# Sprzedaż produktów w 4 sklepach przez 3 dni
sales = np.array([
[100, 150, 200], # Sklep 1
[120, 180, 220], # Sklep 2
[90, 140, 190], # Sklep 3
[110, 160, 210] # Sklep 4
])
print("Sprzedaż (sklepy × dni):")
print(sales)
print("Shape:", sales.shape) # (4, 3)

# Dodajemy bonus 10, 20, 30 do każdego dnia
daily_bonus = np.array([10, 20, 30])
print("\nBonus dzienny:", daily_bonus)
print("Shape:", daily_bonus.shape) # (3,)

# Broadcasting: (4, 3) + (3,) → (4, 3)
sales_with_bonus = sales + daily_bonus
print("\nSprzedaż z bonusem:")
print(sales_with_bonus)

# Inny przykład: mnożnik per sklep
shop_multiplier = np.array([1.1, 1.2, 0.9, 1.0]).reshape(4, 1) # (4, 1)
print("\nMnożnik per sklep:")
print(shop_multiplier)
print("Shape:", shop_multiplier.shape)

# Broadcasting: (4, 3) * (4, 1) → (4, 3)
adjusted_sales = sales * shop_multiplier
print("\nSprzedaż po korekcie:")
print(adjusted_sales)
```

```python
import numpy as np
# Dataset: wzrost [cm] i waga [kg] dla 100 osób
np.random.seed(42)
heights = np.random.normal(170, 10, 100)
weights = np.random.normal(70, 15, 100)
print("Heights shape:", heights.shape) # (100,)
print("Weights shape:", weights.shape) # (100,)

# Standaryzacja (z-score): (x - mean) / std
heights_mean = np.mean(heights)
heights_std = np.std(heights)
heights_standardized = (heights - heights_mean) / heights_std
print("\nWysokości standaryzowane (pierwsze 5):", heights_standardized[:5])
print("Średnia po standaryzacji:", np.mean(heights_standardized)) # ~0
print("Std po standaryzacji:", np.std(heights_standardized)) # ~1

# Batch normalization dla macierzy: 50 próbek × 4 cechy
data = np.random.randn(50, 4) * 10 + 50
print("\n--- Batch Normalization ---")
print("Data shape:", data.shape) # (50, 4)

# Średnia i std per cecha (axis=0)
mean_per_feature = np.mean(data, axis=0) # (4,)
std_per_feature = np.std(data, axis=0) # (4,)
print("Mean per feature:", mean_per_feature)
print("Std per feature:", std_per_feature)

# Standaryzacja z broadcastingiem: (50, 4) - (4,) / (4,)
data_normalized = (data - mean_per_feature) / std_per_feature
print("\nPo normalizacji — średnie per feature:")
print(np.mean(data_normalized, axis=0)) # ~[0, 0, 0, 0]
print("Po normalizacji — std per feature:")
print(np.std(data_normalized, axis=0)) # ~[1, 1, 1, 1]

# Broadcasting oszczędza pamięć!
# Zamiast powielać mean_per_feature 50 razy → (50, 4)
# NumPy "rozciąga" (4,) do (50, 4) wirtualnie podczas obliczeń
```

```python
import time
import numpy as np
n = 1_000_000
# Dane: lista Pythonowa vs tablica NumPy
lista_a = list(range(n))
lista_b = list(range(n))
arr_a = np.arange(n)
arr_b = np.arange(n)

# Wariant 1: czysty Python — pętla + list comprehension
durations_py = []
for _ in range(3):
  start = time.perf_counter()
  wynik_py = [a * b for a, b in zip(lista_a, lista_b)]
  suma_py = sum(wynik_py)
  durations_py.append(time.perf_counter() - start)

# Wariant 2: NumPy — wektoryzacja
durations_np = []
for _ in range(3):
  start = time.perf_counter()
  wynik_np = arr_a * arr_b # operacja element-wise, wektorowo
  suma_np = wynik_np.sum()
  durations_np.append(time.perf_counter() - start)
t_py = sum(durations_py) / len(durations_py)
t_np = sum(durations_np) / len(durations_np)
print(f"Czysty Python (lista) : {t_py:.4f} s")
print(f"NumPy (ndarray) : {t_np:.4f} s")
print(f"Przyspieszenie : {t_py / t_np:.1f}×")
print(f"Zgodność wyników : {suma_py == suma_np}")

# Typowy wynik: NumPy 20-100× szybszy (zależnie od CPU i dtype)
```

```python
import numpy as np
# Normalizacja Min-Max: (x - min) / (max - min) → [0, 1]
min_vals = np.min(iris_data, axis=0) # (4,) — min per cecha
max_vals = np.max(iris_data, axis=0) # (4,)

# Broadcasting: (30, 4) - (4,) / (4,)
iris_normalized = (iris_data - min_vals) / (max_vals - min_vals)
print("Dane po normalizacji (pierwsze 5):")
print(iris_normalized[:5])
print("\nMin per feature:", np.min(iris_normalized, axis=0)) # ~[0, 0, 0, 0]
print("Max per feature:", np.max(iris_normalized, axis=0)) # ~[1, 1, 1, 1]

# Standaryzacja: (x - mean) / std → mean=0, std=1
mean_vals = np.mean(iris_data, axis=0)
std_vals = np.std(iris_data, axis=0)
iris_standardized = (iris_data - mean_vals) / std_vals
print("\n--- Po standaryzacji ---")
print("Mean per feature:", np.mean(iris_standardized, axis=0))
print("Std per feature:", np.std(iris_standardized, axis=0))

# Filtrowanie: Tylko próbki z Petal Length > 4.0
mask = iris_data[:, 2] > 4.0
filtered_data = iris_data[mask]
filtered_labels = iris_labels[mask]
print(f"\nPróbki z Petal Length > 4.0: {filtered_data.shape[0]}/{iris_data.shape[0]}")
print("Gatunki w filtrowanych danych:", np.unique(filtered_labels))
```

```python
import numpy as np
# Histogram za pomocą np.histogram (bez matplotlib)
feature_names = ['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width']
species_names = ['Setosa ', 'Versicolor', 'Virginica ']
print("=== Histogramy cech per gatunek (█ = 1 próbka) ===\n")
for feat_idx, feat_name in enumerate(feature_names):
    print(f"--- {feat_name} ---")
    # Wspólne krawędzie binów dla 3 gatunków
    feature_min = np.min(iris_data[:, feat_idx])
    feature_max = np.max(iris_data[:, feat_idx])
    bins = np.linspace(feature_min, feature_max, 6) # 5 przedziałów
    for species_idx in range(3):
        species_mask = iris_labels == species_idx
        values = iris_data[species_mask, feat_idx]
        counts, _ = np.histogram(values, bins=bins)
        bars = "".join("█" * c for c in counts)
        print(f" {species_names[species_idx]}: {bars} counts={counts.tolist()}")
    print()

# Macierz korelacji (ręczna implementacja — tylko NumPy)
print("=== Macierz korelacji cech Iris ===")
n_features = iris_data.shape[1]
corr_matrix = np.zeros((n_features, n_features))
for i in range(n_features):
    for j in range(n_features):
        if i == j:
            corr_matrix[i, j] = 1.0
        else:
            x = iris_data[:, i]
            y = iris_data[:, j]
            corr = np.sum((x - np.mean(x)) * (y - np.mean(y))) / (len(x) * np.std(x) * np.std(y))
            corr_matrix[i, j] = corr

# Sformatowany wydruk macierzy
header = " " + " ".join(f"{n[:10]:>10s}" for n in feature_names)
print(header)
for i, row_name in enumerate(feature_names):
    row = f"{row_name[:14]:<14s} " + " ".join(f"{corr_matrix[i, j]:+10.2f}" for j in range(n_features))
    print(row)

# Interpretacja
print("\nInterpretacja:")
print("- Wartości bliskie +1 → silna dodatnia korelacja")
print("- Wartości bliskie -1 → silna ujemna korelacja")
print("- Wartości bliskie 0 → brak zależności liniowej")
```

```python
# Zadanie 2 — Operacje wektorowe
import numpy as np

before = np.array([100, 250, 75, 320, 180, 450, 90])
after = np.array([80, 200, 60, 280, 150, 400, 75])

# Oszczędności (różnica cen)
savings = before - after

# Procent rabatu dla każdego produktu
discount_pct = np.round((savings / before) * 100, 2)

# Średnia oszczędność
mean_savings = np.round(np.mean(savings), 2)

print("Ceny przed rabatem:", before)
print("Ceny po rabacie:   ", after)
print("\nOszczędności:", savings)
print("Procent rabatu [%]:", discount_pct)
print("Średnia oszczędność:", mean_savings)

print("\nPodsumowanie:")
print(f"Średnio oszczędzasz {mean_savings:.2f} na produkt; "
      f"największy rabat to {np.max(discount_pct):.2f}%, najmniejszy {np.min(discount_pct):.2f}%.")
```

```python
# Zadanie 3 — Indeksowanie i slicing
import numpy as np

temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])

# 1. Temperatury z dni parzystych (indeksy 0, 2, 4, ...)
even_days = temps[::2]

# 2. Ostatnie 5 temperatur
last_5 = temps[-5:]

# 3. Odwrotna kolejność
reversed_temps = temps[::-1]

# 4. Temperatury większe niż 20 (maska boolean)
mask = temps > 20
above_20 = temps[mask]

print("Wszystkie temperatury:", temps)
print("\n1. Dni parzyste (indeksy 0, 2, 4, ...):", even_days)
print("2. Ostatnie 5 temperatur:", last_5)
print("3. Odwrotna kolejność:", reversed_temps)
print("4. Temperatury > 20:", above_20)
print("   Indeksy tych dni:", np.where(mask)[0])

print("\nPodsumowanie:")
print(f"{above_20.size} z {temps.size} dni miało temperaturę powyżej 20°C.")
```

```python
# Zadanie 5 — Statystyki per wiersz i kolumna
import numpy as np

np.random.seed(1)
sales = np.random.randint(10, 100, size=(4, 7))

# Wiersze = produkty (4), kolumny = dni (7)
mean_per_product = np.mean(sales, axis=1)
mean_per_day = np.mean(sales, axis=0)

total_per_day = np.sum(sales, axis=0)
best_day = np.argmax(total_per_day)

worst_product = np.argmin(mean_per_product)

print("Sprzedaż (wiersze = produkty, kolumny = dni):\n", sales)
print("\nŚrednia sprzedaż per produkt (axis=1):", np.round(mean_per_product, 2))
print("Średnia sprzedaż per dzień (axis=0):", np.round(mean_per_day, 2))
print("\nŁączna sprzedaż per dzień:", total_per_day)
print("Dzień z najwyższą łączną sprzedażą: dzień", best_day + 1,
      f"(indeks {best_day}, suma {total_per_day[best_day]})")
print("Produkt z najniższą średnią sprzedażą: produkt", worst_product + 1,
      f"(indeks {worst_product}, średnia {mean_per_product[worst_product]:.2f})")

print("\nPodsumowanie:")
print(f"Najlepszy dzień to dzień {best_day + 1}, a najsłabiej sprzedaje się produkt {worst_product + 1}.")
```

```python
# Zadanie 6 — Boolean indexing
import numpy as np

grades = np.array([78, 45, 92, 67, 34, 88, 56, 91, 23, 85, 73, 95])

# 1. Oceny wysokie (>= 80)
high = grades[grades >= 80]

# 2. Oceny niedostateczne (< 50)
failing = grades[grades < 50]

# 3. Liczba ocen w przedziale [60, 80)
in_range = (grades >= 60) & (grades < 80)
count_range = np.sum(in_range)

# 4. Podniesienie ocen < 50 do 50 (na kopii, żeby zachować oryginał)
raised = grades.copy()
raised[raised < 50] = 50

print("Oceny:", grades)
print("\n1. Oceny >= 80:", high)
print("2. Oceny < 50:", failing)
print("3. Liczba ocen w przedziale [60, 80):", count_range)
print("   Te oceny:", grades[in_range])
print("4. Oceny po podniesieniu do minimum 50:", raised)
print("   Oryginał bez zmian:", grades)

print("\nPodsumowanie:")
print(f"Wysokich ocen: {high.size}, niedostatecznych: {failing.size}, "
      f"w przedziale [60, 80): {count_range}, "
      f"podniesiono {np.sum(grades < 50)} ocen do 50.")
```

```python
# Zadanie 8 — Reshape i flatten
import numpy as np

arr = np.arange(1, 13)
print("Oryginał:", arr, "shape:", arr.shape)

# 1-3. Zmiany kształtu
a34 = arr.reshape(3, 4)
a26 = arr.reshape(2, 6)
a43 = arr.reshape(4, 3)

print("\n(3, 4):\n", a34)
print("\n(2, 6):\n", a26)
print("\n(4, 3):\n", a43)

# 4. Spłaszczenie do 1D
flat_f = a34.flatten()
flat_r = a34.ravel()
print("\nflatten():", flat_f, "shape:", flat_f.shape)
print("ravel():  ", flat_r, "shape:", flat_r.shape)

# Różnica: kopia vs widok
print("\nCzy flatten() dzieli pamięć z a34?", np.shares_memory(a34, flat_f))
print("Czy ravel()   dzieli pamięć z a34?", np.shares_memory(a34, flat_r))

work = arr.copy().reshape(3, 4)   # osobna tablica do eksperymentu
f = work.flatten()
f[0] = 100
print("\nPo zmianie flatten()[0] = 100, work[0, 0] =", work[0, 0], "(bez zmian)")

r = work.ravel()
r[0] = 100
print("Po zmianie ravel()[0]   = 100, work[0, 0] =", work[0, 0], "(zmienione)")

# Reshape też zwraca widok (tu: a34 dzieli pamięć z arr)
print("\nCzy reshape() zwraca widok? a34 dzieli pamięć z arr:", np.shares_memory(arr, a34))

# Uwaga: ravel() nie zawsze zwraca widok
t = a34.T
print("Czy ravel() na transpozycji dzieli pamięć?", np.shares_memory(t, t.ravel()))

print("\nPodsumowanie:")
print("flatten() zawsze zwraca kopię, ravel() zwykle widok (gdy dane leżą w pamięci ciągle).")
```
