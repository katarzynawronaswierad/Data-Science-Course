```python
import math # moduł z dodatkowymi funkcjami matematycznymi
# Obliczanie podstawowych statystyk (później użyjemy NumPy/Pandas)
# Przykład: ręczne obliczenie średniej dla 5 pomiarów temperatury
temp_1 = 22.3
temp_2 = 23.1
temp_3 = 21.8
temp_4 = 24.2
temp_5 = 22.6
# Średnia arytmetyczna
suma_temp = temp_1 + temp_2 + temp_3 + temp_4 + temp_5
liczba_pomiarow = 5
srednia = suma_temp / liczba_pomiarow
print(f"Średnia temperatura: {srednia:.2f}°C") # 22.80°C
```

```python
# Obliczanie wariancji (uproszczone, bez NumPy)
# Wariancja = średnia kwadratów odchyleń od średniej
odch_1 = (temp_1 - srednia) ** 2
odch_2 = (temp_2 - srednia) ** 2
odch_3 = (temp_3 - srednia) ** 2
odch_4 = (temp_4 - srednia) ** 2
odch_5 = (temp_5 - srednia) ** 2
wariancja = (odch_1 + odch_2 + odch_3 + odch_4 + odch_5) / liczba_pomiarow
odchylenie_std = math.sqrt(wariancja) # pierwiastek z wariancji
```

```python
#Zadanie 3. Podział zbioru danych
total = 1000
pct_train = 70
pct_vali = 15
pct_test = 15

# Mnożenie, a potem dzielenie całkowite (kolejność ma znaczenie!)
train = total * pct_train // 100
vali = total * pct_vali // 100

#Zbiór testowy dostaje wszystko, w tym resztę z dzielenia
test = total - train - vali

#Wyświetl nominalne wielkości zbiorów
print(f"Zbiór treningowy:   {train}")
print(f"Zbiór walidacyjny:  {vali}")
print(f"Zbiór testowy:      {test}")
print(f"Suma:               {train + vali + test}")

#assert zatrzyma program z błędem, gdyby suma przy zmianie proporcji się nie zgadzała

assert train + vali + test == total
```

```python
#Zadanie 5. Parsing ścieżki pliku
path = "data/processed/train_fold_03_normalized.csv"

# Podział ścieżki
parts = path.split("/")

folder_1 = parts[0]  # data
folder_2 = parts[1]  # processed

# Nazwa pliku bez rozszerzenia
filename_parts = parts[2].split("_")

dataset_type = filename_parts[0]  # train
fold_number = filename_parts[2]   # 03

# Format pliku
file_format = parts[2].split(".")[1]  # csv

print("Folder 1:", folder_1)
print("Folder 2:", folder_2)
print("Typ zbioru:", dataset_type)
print("Numer fold:", fold_number)
print("Format pliku:", file_format)
```

```python
#Zadanie 6. Dokładność w procentach
accuracy = 0.9234
precision = 0.8891
recall= 0.9456

#Wyświetl każdą metrykę w procentach z 2 miejscami po przecinku
print("Accuracy: {:.2f}%".format(accuracy * 100))
print("Precision: {:.2f}%".format(precision * 100))
print("Recall: {:.2f}%".format(recall * 100))

#Użyj f-strings z formatowaniem :.2%
print(f"Accuracy: {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")

#Dodaj estetyczne wyrównanie kolumn
print("{:<12} {:>4.2f}%".format("Accuracy", accuracy * 100))
print("{:<12} {:>4.2f}%".format("Precision", precision * 100))
print("{:<12} {:>4.2f}%".format("Recall", recall * 100))

print(f"{'Accuracy:':<12} {accuracy:.2%}")
print(f"{'Precision:':<12} {precision:.2%}")
print(f"{'Recall:':<12} {recall:.2%}")
```

```python
#Zadanie 7. Porównanie dwóch modeli

accuracy_a = 0.87
train_time_a = 120

accuracy_b = 0.89
train_time_b = 450

accuracy_diff = accuracy_b - accuracy_a
time_diff = train_time_b - train_time_a

print(f"Różnica accuracy: {accuracy_diff:.2f}")
print(f"Różnica czasu treningu: {time_diff} s")

# Najlepszy kompromis
if (accuracy_diff > 0.02) and (accuracy_b > accuracy_a):
    recommendation = "Model B - wyższa accuracy uzasadnia dłuższy trening"

elif (accuracy_diff <= 0.02) and (train_time_a < train_time_b):
    recommendation = "Model A - podobna accuracy, znacznie krótszy trening"

else:
    recommendation = "Brak jednoznacznego zwycięzcy"

print("\nRekomendacja:")
print(recommendation)
```

```python
#Zadanie 8. Obliczanie learning rate dla schedulera

# Learning rate scheduler: zanik wykładniczy (exponential decay)

lr_initial = 0.1
decay = 0.95
epoch = 10

lr_new = lr_initial * (decay ** epoch)

print(f"Learning rate po {epoch} epokach : {lr_new:.2e}")
```
