```python
# Confusion matrix 2x2 dla klasyfikacji binarnej
confusion_matrix = [
[50, 5], # wiersz 0: TN (True Negative), FP (False Positive)
[3, 42] # wiersz 1: FN (False Negative), TP (True Positive)
]
print(f"Confusion Matrix:")
print(f" {confusion_matrix[0]}")
print(f" {confusion_matrix[1]}")
```

```python
# Indeksowanie dwuwymiarowe: [wiersz][kolumna]
tn = confusion_matrix[0][0]
fp = confusion_matrix[0][1]
fn = confusion_matrix[1][0]
tp = confusion_matrix[1][1]
print(f"TN = {tn}, FP = {fp}, FN = {fn}, TP = {tp}")
```

```python
# Obliczenia bezpośrednio na elementach listy
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp)
recall = tp / (tp + fn)
print(f"Accuracy : {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall : {recall:.2%}")

```

```python
import time
# Lista 5000 fragmentów
czesci = ['fragment'] * 5000
# Wariant NIEEFEKTYWNY – każde '+' tworzy nowy string
start = time.time()
wynik_bad = ""
wynik_bad = wynik_bad + czesci[0]
wynik_bad = wynik_bad + czesci[1]
wynik_bad = wynik_bad + czesci[2]
# (powtórzone 5000 razy wymagałoby pętli – w L7 zobaczysz, jak to zmierzyć)
# Tu demonstrujemy sam mechanizm:
czas_bad = time.time() - start
# Wariant EFEKTYWNY – jedno wywołanie join
start = time.time()
wynik_good = ''.join(czesci)
czas_good = time.time() - start
print(f"Długość wyniku join: {len(wynik_good)} znaków")
print(f"Czas join: {czas_good:.6f} s")
```

```python
# Jedna linia w formacie CSV z datasetu Iris
# Kolejność: sepal_length, sepal_width, petal_length, petal_width, species
linia = "5.1,3.5,1.4,0.2,setosa"
# Krok 1: rozdzielenie po przecinku
pola = linia.split(',')
print(f"Pola po split: {pola}")
# Krok 2: konwersja typów
sepal_length = float(pola[0])
sepal_width = float(pola[1])
petal_length = float(pola[2])
petal_width = float(pola[3])
species = pola[4]
# Krok 3: obliczenia na pojedynczym rekordzie
sepal_ratio = sepal_length / sepal_width
petal_ratio = petal_length / petal_width
sepal_area = sepal_length * sepal_width # przybliżona "powierzchnia"
# Krok 4: sformatowany raport
print("-" * 40)
print(f"Gatunek: {species}")
print(f"Działka kielicha: {sepal_length} × {sepal_width} cm")
print(f"Płatek: {petal_length} × {petal_width} cm")
print(f"Proporcja działki: {sepal_ratio:.2f}")
print(f"Proporcja płatka: {petal_ratio:.2f}")
print(f"Pseudo-powierzchnia: {sepal_area:.2f} cm²")
print("-" * 40)
```

```python
# Ta sama linia, ale "brudna" – tak, jak bywa w realnych danych
brudna = " 5.1 , 3.5 ,1.4, 0.2, SETOSA "
```

```python
# Krok 1: strip globalny (usuwa spacje z początku/końca całej linii)
bez_spacji = brudna.strip()
print(f"Po strip: '{bez_spacji}'")
```

```python
# Krok 2: split
pola = bez_spacji.split(',')
print(f"Po split: {pola}")
```

```python
# Krok 3: strip każdego pola indywidualnie (bez pętli – z indeksem)
pole_0 = pola[0].strip()
pole_1 = pola[1].strip()
pole_2 = pola[2].strip()
pole_3 = pola[3].strip()
pole_4 = pola[4].strip().lower() # kategorię też normalizujemy do małych liter
print(f"Po czyszczeniu: ['{pole_0}', '{pole_1}', '{pole_2}', '{pole_3}', '{pole_4}']")
```

```python
# Krok 4: konwersja na typy numeryczne
sepal_length = float(pole_0)
sepal_width = float(pole_1)
petal_length = float(pole_2)
petal_width = float(pole_3)
species = pole_4
print(f"Odczytano: {species}, sepal=({sepal_length}, {sepal_width}), petal=({petal_length}, {petal_width})")
```
