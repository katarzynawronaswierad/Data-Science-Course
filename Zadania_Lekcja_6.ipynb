```python
#Zadanie 5. Operacje na dwóch zbiorach
grupa_a = {'python', 'pandas', 'numpy', 'sql', 'tableau'}
grupa_b = {'python', 'r', 'sql', 'ggplot2', 'dplyr'}

# 1. Suma zbiorów
print(f"Suma (operator): {grupa_a | grupa_b}")
print(f"Suma (metoda):   {grupa_a.union(grupa_b)}")

# 2. Wspólne narzędzia
print(f"Iloczyn (operator): {grupa_a & grupa_b}")
print(f"Iloczyn (metoda):   {grupa_a.intersection(grupa_b)}")

# 3. Narzędzia tylko w grupie A
print(f"Tylko A (operator): {grupa_a - grupa_b}")
print(f"Tylko A (metoda):   {grupa_a.difference(grupa_b)}")

# 4. Narzędzia tylko w jednej z grup
print(f"Różnica symetryczna (operator): {grupa_a ^ grupa_b}")
print(f"Różnica symetryczna (metoda):   {grupa_a.symmetric_difference(grupa_b)}")

# 5. Czy zbiory są rozłączne?
print(f"Rozłączne: {grupa_a.isdisjoint(grupa_b)}")
```

```python
#Zadanie 7. Rozpakowanie rekordu
pracownik = ('Julia', 'Wiśniewska', 28, 'Data Analyst', 8500.0)

# 1. Rozpakowanie wszystkich pól
imie, nazwisko, wiek, stanowisko, pensja = pracownik

# 2. Wypisanie f-stringiem
print(f"{imie} {nazwisko} ({wiek}) – {stanowisko}, pensja: {pensja:.2f} zł")

# 3. Rozpakowanie z gwiazdką
imie, *reszta = pracownik

# 4. Wypisanie imienia i reszty
print(imie)
print(reszta)
```

```python
#Zadanie 8. Krotka jako klucz słownika
sprzedaz = {
    (2024, 'Q1'): 120_000,
    (2024, 'Q2'): 140_000,
    (2024, 'Q3'): 135_000,
    (2024, 'Q4'): 160_000
}

# 1. Sprzedaż z (2024, 'Q2')
print(sprzedaz[(2024, 'Q2')])

# 2. Dodanie wpisu
sprzedaz[(2025, 'Q1')] = 130_000

# 3. Najwyższa, najniższa i łączna sprzedaż
print(f"Najwyższa: {max(sprzedaz.values())}")
print(f"Najniższa: {min(sprzedaz.values())}")
print(f"Łączna: {sum(sprzedaz.values())}")

# 4. Liczba wpisów (różnych par rok-kwartał)
print(len(sprzedaz))
```
