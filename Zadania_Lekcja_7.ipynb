```python
# Symulacja zbieżności algorytmu
wartosc = 100.0
prog = 0.01
iteracja = 0
max_iteracji = 1000

while True: # pętla „nieskończona" z warunkiem wyjścia w środku
  iteracja += 1
  nowa = wartosc * 0.95 # krok algorytmu
  zmiana = abs(wartosc - nowa)

  if zmiana < prog:
    print(f"Zbieżność w iteracji {iteracja}, wartość = {nowa:.4f}")
    break

  if iteracja >= max_iteracji:
    print(f"Limit iteracji ({max_iteracji}) – przerwano.")
    break

  wartosc = nowa

print(f"Iteracji łącznie: {iteracja}")
```

```python
miasta = ['Warszawa', 'Kraków', 'Gdańsk', 'Wrocław', 'Poznań']
wspolrzedne = [
(52.23, 21.01), # Warszawa
(50.06, 19.94), # Kraków
(54.35, 18.65), # Gdańsk
(51.11, 17.04), # Wrocław
(52.41, 16.93) # Poznań
]

# Nagłówek tabeli
print(" ", end='')
for miasto in miasta:
  print(f"{miasto:>10}", end='')
print()

# Wiersze macierzy – dwie pętle for w range
for i in range(len(miasta)):
  print(f"{miasta[i]:8}", end='')
  for j in range(len(miasta)):
    if i == j:
      odl = 0.0
    else:
      lat1, lon1 = wspolrzedne[i]
      lat2, lon2 = wspolrzedne[j]

      # Euklides w stopniach * 111 (~ km na 1° szerokości)
      odl = ((lat2 - lat1) ** 2 + (lon2 - lon1) ** 2) ** 0.5 * 111
    print(f"{odl:>10.1f}", end='')
  print()
```

```python
min_odl = 9999999.0
max_odl = 0.0
para_min = ('', '')
para_max = ('', '')

for i in range(len(miasta)):
  # j zaczyna się od i+1 – dzięki temu każdą parę liczymy raz
  for j in range(i + 1, len(miasta)):
    lat1, lon1 = wspolrzedne[i]
    lat2, lon2 = wspolrzedne[j]
    odl = ((lat2 - lat1) ** 2 + (lon2 - lon1) ** 2) ** 0.5 * 111

  if odl < min_odl:
    min_odl = odl
    para_min = (miasta[i], miasta[j])

  if odl > max_odl:
    max_odl = odl
    para_max = (miasta[i], miasta[j])

print(f"Najbliższe: {para_min[0]} – {para_min[1]} ({min_odl:.1f} km)")
print(f"Najdalsze: {para_max[0]} – {para_max[1]} ({max_odl:.1f} km)")
```

```python
# 10 pierwszych rekordów datasetu Iris (uproszczony)
iris = [
  {'gatunek': 'setosa', 'dl_platka': 1.4, 'sz_platka': 0.2},
  {'gatunek': 'setosa', 'dl_platka': 1.4, 'sz_platka': 0.2},
  {'gatunek': 'setosa', 'dl_platka': 1.3, 'sz_platka': 0.2},
  {'gatunek': 'setosa', 'dl_platka': 1.5, 'sz_platka': 0.2},
  {'gatunek': 'versicolor', 'dl_platka': 4.7, 'sz_platka': 1.4},
  {'gatunek': 'versicolor', 'dl_platka': 4.5, 'sz_platka': 1.5},
  {'gatunek': 'versicolor', 'dl_platka': 4.9, 'sz_platka': 1.5},
  {'gatunek': 'virginica', 'dl_platka': 6.0, 'sz_platka': 2.5},
  {'gatunek': 'virginica', 'dl_platka': 5.1, 'sz_platka': 1.9},
  {'gatunek': 'virginica', 'dl_platka': 5.9, 'sz_platka': 2.1}
]

# --- 1. Eksploracja ---
print("Rekordy:")
for i, k in enumerate(iris, start=1):
  print(f" #{i:2}: {k['gatunek']:12} "
        f"dł={k['dl_platka']:.1f}, sz={k['sz_platka']:.1f}")

# --- 2. Unikalne gatunki (zbiór z comprehension) ---
gatunki_unikalne = set([k['gatunek'] for k in iris])
print(f"\nUnikalne gatunki: {gatunki_unikalne}")

# --- 3. Statystyki per gatunek ---
print("\n=== Statystyki per gatunek ===")
for g in sorted(gatunki_unikalne):
  dlugosci = [k['dl_platka'] for k in iris if k['gatunek'] == g]
  szerokosci = [k['sz_platka'] for k in iris if k['gatunek'] == g]
  print(f"\n{g}:")
  print(f" liczba próbek: {len(dlugosci)}")
  print(f" śr. długość płatka: {sum(dlugosci) / len(dlugosci):.2f}")
  print(f" śr. szerokość płatka: {sum(szerokosci) / len(szerokosci):.2f}")

# --- 4. Prosta klasyfikacja rule-based ---
# Reguła: dl < 2 -> setosa; 2 <= dl < 5 -> versicolor; dl >= 5 -> virginica
print("\n=== Prosta klasyfikacja rule-based ===")
poprawne = 0
for i, k in enumerate(iris, start=1):
  dl = k['dl_platka']
  if dl < 2:
    predykcja = 'setosa'
  elif dl < 5:
    predykcja = 'versicolor'
  else:
    predykcja = 'virginica'

  prawda = k['gatunek']
  ok = predykcja == prawda
  if ok:
    poprawne += 1
  status = '✓' if ok else '✗'
  print(f" {status} #{i:2} dl={dl:.1f} -> pred={predykcja:12} "
        f"| true={prawda}")

accuracy = poprawne / len(iris) * 100
print(f"\nAccuracy: {poprawne}/{len(iris)} = {accuracy:.1f}%")
```
