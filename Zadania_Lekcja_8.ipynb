```python
# Funkcja z wieloma parametrami do czyszczenia danych
def przygotuj_dane(lista_wartosci, min_wartosc, max_wartosc):
  """
  Filtruje listę wartości, pozostawiając tylko te w zadanym zakresie.
  Args:
  lista_wartosci (list): Lista liczb do przefiltrowania
  min_wartosc (float): Minimalna akceptowalna wartość
  max_wartosc (float): Maksymalna akceptowalna wartość
  Returns:
  list: Przefiltrowana lista wartości
  """
  czyste_dane = []
  for wartosc in lista_wartosci:
    if min_wartosc <= wartosc <= max_wartosc:
      czyste_dane.append(wartosc)
  return czyste_dane

# Dane z czujników temperatury (niektóre błędne)
temperatury = [22.1, 23.5, -999, 24.2, 25.1, 100, 23.8, 22.9]

# Filtrowanie danych - akceptujemy tylko temperatury 15-35°C
temperatury_ok = przygotuj_dane(temperatury, 15, 35)
print(f"Oryginalne dane: {len(temperatury)} pomiarów")
print(f"Po czyszczeniu: {len(temperatury_ok)} pomiarów")
print(f"Czyste dane: {temperatury_ok}")
# Wynik: [22.1, 23.5, 24.2, 25.1, 23.8, 22.9]
```

```python
# Funkcja z parametrami domyślnymi
def normalizuj_dane(dane, metoda="min-max", zaokraglenie=2):
  """
  Normalizuje dane przy użyciu wybranej metody.
  Args:
  dane (list): Lista wartości do normalizacji
  metoda (str): Metoda normalizacji ("min-max" lub "z-score")
  zaokraglenie (int): Liczba miejsc po przecinku
  Returns:
  list: Znormalizowane dane
  """
  if metoda == "min-max":
    # Normalizacja min-max: (x - min) / (max - min)
    min_val = min(dane)
    max_val = max(dane)
    zakres = max_val - min_val

    znormalizowane = []
    for wartosc in dane:
      norma = (wartosc - min_val) / zakres
      znormalizowane.append(round(norma, zaokraglenie))
    return znormalizowane

  elif metoda == "z-score":
    # Normalizacja z-score: (x - średnia) / odchylenie standardowe
    # Obliczamy średnią
    srednia = sum(dane) / len(dane)

    # Obliczamy odchylenie standardowe
    suma_kwadratow = sum((x - srednia) ** 2 for x in dane)
    odchylenie = (suma_kwadratow / len(dane)) ** 0.5

    znormalizowane = []
    for wartosc in dane:
      z = (wartosc - srednia) / odchylenie
      znormalizowane.append(round(z, zaokraglenie))
    return znormalizowane

# Dane - ceny produktów w PLN
ceny = [49.99, 89.99, 129.99, 199.99, 299.99]
# Normalizacja min-max (domyślna)
ceny_norm_minmax = normalizuj_dane(ceny)
print(f"Oryginalne ceny: {ceny}")
print(f"Normalizacja min-max: {ceny_norm_minmax}")
# Wynik: [0.0, 0.16, 0.32, 0.6, 1.0]

# Normalizacja z-score
ceny_norm_zscore = normalizuj_dane(ceny, metoda="z-score", zaokraglenie=3)
print(f"Normalizacja z-score: {ceny_norm_zscore}")
# Wynik: [-1.179, -0.725, -0.272, 0.521, 1.655]
```

```python
# Funkcja zwracająca tuple z wieloma wynikami
def statystyki_opisowe(dane):
  """
  Oblicza podstawowe statystyki opisowe dla danych.

  Args:
  dane (list): Lista wartości numerycznych

  Returns:
  tuple: Średnia, mediana, minimum, maksimum, zakres
  """
  # Średnia
  srednia = sum(dane) / len(dane)

  # Mediana (sortujemy i bierzemy środkowy element)
  posortowane = sorted(dane)
  n = len(posortowane)
  if n % 2 == 0:
    mediana = (posortowane[n//2 - 1] + posortowane[n//2]) / 2
  else:
    mediana = posortowane[n//2]

  # Minimum i maksimum
  minimum = min(dane)
  maksimum = max(dane)

  # Zakres (rozstęp)
  zakres = maksimum - minimum

  # Zwracamy wszystkie wyniki jako tuple
  return srednia, mediana, minimum, maksimum, zakres

# Dane - dzienny czas pracy w godzinach
czas_pracy = [8.5, 7.2, 9.1, 8.0, 6.5, 10.2, 7.8, 8.3, 9.5, 7.9]

# Rozpakowywanie zwróconych wartości (tuple unpacking)
sr, med, mini, maxi, zak = statystyki_opisowe(czas_pracy)

print("Statystyki czasu pracy:")
print(f" Średnia: {sr:.1f}h")
print(f" Mediana: {med:.1f}h")
print(f" Minimum: {mini:.1f}h")
print(f" Maksimum: {maxi:.1f}h")
print(f" Zakres: {zak:.1f}h")
```

```python
# Dane klientów - wiek, dochód, wydatki miesięczne
klienci = [
{"id": 1, "wiek": 25, "dochod": 4500, "wydatki": 3200},
{"id": 2, "wiek": 35, "dochod": 7800, "wydatki": 5100},
{"id": 3, "wiek": 45, "dochod": 9200, "wydatki": 4800},
{"id": 4, "wiek": 28, "dochod": 5200, "wydatki": 4900},
{"id": 5, "wiek": 52, "dochod": 11000, "wydatki": 6200}
]

# Feature engineering - tworzymy nową cechę: wskaźnik oszczędności
# Oszczędności = (dochód - wydatki) / dochód
dodaj_wskaznik_oszczednosci = lambda k: {
  **k, # rozpakowujemy istniejące pola słownika (z L6)
  "wskaznik_oszczednosci": round((k["dochod"] - k["wydatki"]) / k["dochod"], 2)
}

klienci_z_wskaznikiem = list(map(dodaj_wskaznik_oszczednosci, klienci))

print("Klienci z wskaźnikiem oszczędności:")
for klient in klienci_z_wskaznikiem:
  print(f" ID {klient['id']}: dochód={klient['dochod']}, "
    f"wydatki={klient['wydatki']}, "
    f"wskaźnik={klient['wskaznik_oszczednosci']}")

# Segmentacja klientów - wysoki wskaźnik oszczędności (≥0.30)
oszczedni_klienci = list(filter(
  lambda k: k["wskaznik_oszczednosci"] >= 0.30,
  klienci_z_wskaznikiem
))
print(f"\nKlienci oszczędni (wskaźnik ≥30%): {len(oszczedni_klienci)}")
```

```python
# Dane klientów - wiek i miesięczny dochód
klienci_raw = [
(1, 25, 4500), # (id, wiek, dochód)
(2, 35, 7800),
(3, 45, 9200),
(4, 28, 5200),
(5, 52, 11000),
(6, 31, 6800),
(7, 42, 8500),
(8, 29, 5900)
]

# Feature engineering - kategorie wiekowe
def kategoria_wiekowa(wiek):
  if wiek < 30:
    return "mlody"
  elif wiek < 45:
    return "sredni"
  else:
    return "starszy"

# Feature engineering - poziom dochodów
def poziom_dochodu(dochod):
  if dochod < 5000:
    return "niski"
  elif dochod < 8000:
    return "sredni"
  else:
    return "wysoki"

# Tworzymy nowe cechy używając comprehension
klienci_engineered = [
  {
    "id": k[0],
    "wiek": k[1],
    "dochod": k[2],
    "kat_wiekowa": kategoria_wiekowa(k[1]),
    "poz_dochodu": poziom_dochodu(k[2]),
    "dochod_na_rok_zycia": round(k[2] / k[1], 2)
  }
  for k in klienci_raw
]
```

```python
# Dane - transakcje sprzedażowe
transakcje = [
  {"id": 1, "produkt": "laptop", "cena": 2999, "ilosc": 2},
  {"id": 2, "produkt": "mysz", "cena": 89, "ilosc": 5},
  {"id": 3, "produkt": "klawiatura", "cena": 249, "ilosc": 3},
  {"id": 4, "produkt": "monitor", "cena": 899, "ilosc": 1},
  {"id": 5, "produkt": "laptop", "cena": 2999, "ilosc": 1}
]
# Agregacja - suma wartości dla każdego produktu
# Grupowanie przy pomocy zwykłego słownika z dict.get()
wartosci_produktow = {}
for trans in transakcje:
  prod = trans["produkt"]
  wartosc = trans["cena"] * trans["ilosc"]
  wartosci_produktow[prod] = wartosci_produktow.get(prod, 0) + wartosc
print("Całkowita wartość sprzedaży po produktach:")
for produkt, wartosc in wartosci_produktow.items():
  print(f" {produkt}: {wartosc} PLN")

# Set comprehension - unikalne produkty drogie (cena jednostkowa >500)
drogie_produkty_set = {t["produkt"] for t in transakcje if t["cena"] > 500}
print(f"\nUnikalne drogie produkty: {drogie_produkty_set}")

# List comprehension - transakcje o dużej wartości (>1000)
duze_transakcje = [
{"id": t["id"], "produkt": t["produkt"], "wartosc": t["cena"] * t["ilosc"]}
for t in transakcje
if t["cena"] * t["ilosc"] > 1000
]

print(f"\nDuże transakcje (wartość >1000 PLN):")
for trans in duze_transakcje:
  print(f" ID {trans['id']}: {trans['produkt']} - {trans['wartosc']} PLN")
```

```python
import time
n = 1_000_000

# Wariant 1: pętla for + append()
durations_petla = []
for _ in range(5):
  start = time.perf_counter()
  wynik = []
  for i in range(n):
    wynik.append(i * 2)
  durations_petla.append(time.perf_counter() - start)

# Wariant 2: list comprehension
durations_comp = []
for _ in range(5):
  start = time.perf_counter()
  wynik = [i * 2 for i in range(n)]
  durations_comp.append(time.perf_counter() - start)

t_petla = sum(durations_petla) / len(durations_petla)
t_comp = sum(durations_comp) / len(durations_comp)

print(f"Pętla for + append : {t_petla:.4f} s")
print(f"List comprehension : {t_comp:.4f} s")
print(f"Przyspieszenie : {t_petla / t_comp:.2f}×")
# Przykładowy wynik: comprehension jest ~1.2-1.5× szybsza
```

```python
# Dane sprzedażowe - (dzień, kategoria, liczba transakcji, wartość w PLN)
sprzedaz_raw = [
(1, "elektronika", 45, 12500),
(1, "odziez", 78, 4200),
(1, "ksiazki", 34, 890),
(2, "elektronika", 52, 15200),
(2, "odziez", 65, 3800),
(2, "ksiazki", 29, 760),
(3, "elektronika", 38, 9800),
(3, "odziez", 82, 5100),
(3, "ksiazki", 41, 1020),
(4, "elektronika", 61, 18900),
(4, "odziez", 71, 4500),
(4, "ksiazki", 38, 950),
(5, "elektronika", 44, 11200),
(5, "odziez", 88, 5600),
(5, "ksiazki", 45, 1150)
]

# Nazwy kolumn: (dzień, kategoria, liczba_transakcji, wartość_PLN)
```

```python
# Funkcja konwertująca tuple na słownik
def tuple_to_dict(rekord):
  """Konwertuje tuple danych sprzedażowych na słownik."""
  return {
    "dzien": rekord[0],
    "kategoria": rekord[1],
    "liczba_transakcji": rekord[2],
    "wartosc_pln": rekord[3]
  }

# List comprehension - konwersja wszystkich rekordów
sprzedaz_dict = [tuple_to_dict(r) for r in sprzedaz_raw]

print("Przykładowe rekordy po konwersji:")
for rekord in sprzedaz_dict[:3]:
  print(f" Dzień {rekord['dzien']}, {rekord['kategoria']}: "
    f"{rekord['liczba_transakcji']} transakcji, "
    f"{rekord['wartosc_pln']} PLN")
```

```python
# Dodajemy nowe cechy: średnia wartość transakcji
sprzedaz_engineered = [
  {
    **rekord,
    "srednia_wartosc_transakcji": round(
      rekord["wartosc_pln"] / rekord["liczba_transakcji"],
      2
    )
  }
  for rekord in sprzedaz_dict
]
print("\nRekordy z nową cechą 'średnia wartość transakcji':")
for rekord in sprzedaz_engineered[:3]:
  print(f" {rekord['kategoria']}: średnia transakcja = "
    f"{rekord['srednia_wartosc_transakcji']} PLN")
```

```python
# Funkcja agregująca sprzedaż dla kategorii
def agreguj_kategorie(dane, kategoria):
  """
  Oblicza łączne statystyki dla danej kategorii.
  Returns:
  dict: statystyki (suma transakcji, suma wartości, średnia)
  """
  # Filtrujemy dane dla kategorii
  dane_kat = [r for r in dane if r["kategoria"] == kategoria]

  # Agregacje
  suma_transakcji = sum(r["liczba_transakcji"] for r in dane_kat)
  suma_wartosci = sum(r["wartosc_pln"] for r in dane_kat)
  srednia_dzienna = suma_wartosci / len(dane_kat)

  return {
    "kategoria": kategoria,
    "suma_transakcji": suma_transakcji,
    "suma_wartosci": suma_wartosci,
    "srednia_dzienna": round(srednia_dzienna, 2),
    "liczba_dni": len(dane_kat)
  }

# Unikalne kategorie - set comprehension
kategorie = {r["kategoria"] for r in sprzedaz_engineered}

# Agregacja dla każdej kategorii
statystyki_kategorii = [
  agreguj_kategorie(sprzedaz_engineered, kat)
  for kat in kategorie
]

print("\n=== Statystyki sprzedaży po kategoriach ===")
for stat in statystyki_kategorii:
  print(f"\n{stat['kategoria'].upper()}:")
  print(f" Suma transakcji: {stat['suma_transakcji']}")
  print(f" Suma wartości: {stat['suma_wartosci']} PLN")
  print(f" Średnia dzienna: {stat['srednia_dzienna']} PLN")
```

```python
# Sortujemy kategorie według sumy wartości (malejąco)
statystyki_sorted = sorted(
  statystyki_kategorii,
  key=lambda s: s["suma_wartosci"],
  reverse=True
)

print("\n=== Ranking kategorii (według wartości sprzedaży) ===")
for i, stat in enumerate(statystyki_sorted, 1):
  print(f"{i}. {stat['kategoria']}: {stat['suma_wartosci']} PLN")
```

```python
# Zadanie 1 - Kalkulator statystyk
def statystyki(liczby):
    suma = sum(liczby)
    srednia = round(suma / len(liczby), 2)
    return (suma, srednia, min(liczby), max(liczby))

oceny = [3.5, 4.0, 5.0, 3.0, 4.5, 5.0, 4.0, 3.5]
suma, srednia, minimum, maksimum = statystyki(oceny)

print(f"Suma:     {suma}")
print(f"Średnia:  {srednia}")
print(f"Minimum:  {minimum}")
print(f"Maksimum: {maksimum}")
```

```python
# Zadanie 2 - Konwerter jednostek
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5 / 9

def celsius_to_kelvin(c):
    return c + 273.15

temperatury_c = [0, 10, 20, 30, 37, 100]

print(f"{'°C':>6} {'°F':>8} {'K':>8}")
print("-" * 24)
for c in temperatury_c:
    f = celsius_to_fahrenheit(c)
    k = celsius_to_kelvin(c)
    print(f"{c:>6} {f:>8.1f} {k:>8.2f}")
```

```python
# Zadanie 6 - Funkcja z wartościami domyślnymi
def powitanie(imie, jezyk="polski"):
    if jezyk == "polski":
        return f"Cześć, {imie}!"
    elif jezyk == "angielski":
        return f"Hello, {imie}!"
    elif jezyk == "niemiecki":
        return f"Hallo, {imie}!"
    else:
        return f"Nieobsługiwany język: {jezyk}"

print(f'powitanie("Anna") → "{powitanie("Anna")}"')
print(f'powitanie("Anna", "angielski") → "{powitanie("Anna", "angielski")}"')
print(f'powitanie("Anna", "niemiecki") → "{powitanie("Anna", "niemiecki")}"')
print(f'powitanie("Anna", "francuski") → "{powitanie("Anna", "francuski")}"')
```
